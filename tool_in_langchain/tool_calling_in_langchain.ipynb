```python
# tool create
from langchain_core.tools import InjectedToolArg
from typing import Annotated

@tool
def get_conversion_factor(base_currency: str, target_currency: str) -> float:
  """
  This function fetches the currency conversion factor between a given base currency and a target currency
  """
  url = f'https://v6.exchangerate-api.com/v6/c60ba87416c084d5961a3178/pair/{base_currency}/{target_currency}'

  response = requests.get(url)

  return response.json()

@tool
def convert(base_currency_value: int, conversion_rate: Annotated[float, InjectedToolArg]) -> float:
  """
  given a currency conversion rate this function calculates the target currency value from a given base currency value
  """

  return base_currency_value * conversion_rate
```

```python
from langchain.agents import create_agent

agent = create_agent(
    model=llm,
    tools=[get_conversion_factor, convert],
    system_prompt="""
    You are a currency conversion assistant.

    For currency conversion requests:
    1. Call get_conversion_factor first.
    2. Use its returned conversion rate.
    3. Call convert with the amount and retrieved rate.
    4. Give the user the final result.
    Do not invent exchange rates.
    """
)
```

```python
user_query = (
    "What is the conversion factor between USD and NPR, "
    "and convert 10 USD to NPR."
)

response = agent.invoke({
    "messages": [
        {"role": "user", "content": user_query}
    ]
})

print(response["messages"][-1].content)
```

```python
for msg in response["messages"]:
    print("Message type:", type(msg).__name__)

    if getattr(msg, "tool_calls", None):
        print("Tool calls:", msg.tool_calls)
```
